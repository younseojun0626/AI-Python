```python
import random

num = random.randint(1,100)
user = int(input("1이상 100이하의 숫자를 입력하세요:"))

if num == user:
    print("일치합니다")
elif num < user:
    print("입력하신 숫자가 더 큽니다.")
else:
    print("입력하신 숫자가 더 작습니다.")
print(f"정답은 {num}입니다")
```

```python
def add(n1,n2):
    a = random.randint(1,n1)
    b = random.randint(1,n2)
    result = a + b
    print(f"{a} + {b} = {result}")
```

```python
x = int(input("첫 번째 수:"))
y = int(input("두 번째 수:"))
op = input("연산자를 입력하시오")
def asmd(n1,n2,cal):
    if cal == "+":
        result = n1+n2
    elif cal == "*":
        result = n1*n2
    elif cal == "-":
        result = n1-n2
    elif cal == "/":
        result = n1/n2
    print(result)


asmd(x,y,op)
```
